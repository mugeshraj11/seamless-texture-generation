```python
import os
from PIL import Image
```

```python
INPUT_DIR = "convertible_data/data/floors_wrapped"
OUTPUT_DIR = "convertible_data/data/tiled_preview"
```

```python
GRID_SIZE = 8  # 8x8 tiling
TILE_SIZE = 32
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
for fname in sorted(os.listdir(INPUT_DIR)):
    if not fname.endswith(".png"):
        continue

    img = Image.open(os.path.join(INPUT_DIR, fname))

    preview = Image.new(
        "RGB",
        (GRID_SIZE * TILE_SIZE, GRID_SIZE * TILE_SIZE)
    )

    for y in range(GRID_SIZE):
        for x in range(GRID_SIZE):
            preview.paste(img, (x * TILE_SIZE, y * TILE_SIZE))

    preview.save(os.path.join(OUTPUT_DIR, fname))
```
