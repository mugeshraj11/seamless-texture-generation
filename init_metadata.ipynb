```python
import os
import csv
```

```python
PROCESSED_DIR = "Textures/floor/processed"
METADATA_PATH = "Textures/floor/metadata.csv"
```

```python
with open(METADATA_PATH, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["filename", "seamless"])  # seamless: 1=yes, 0=no

    for fname in sorted(os.listdir(PROCESSED_DIR)):
        if fname.endswith(".png"):
            writer.writerow([fname, ""])  # fill manually later
```
