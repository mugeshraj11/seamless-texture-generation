```python
import os
from PIL import Image
```

```python
RAW_DIR = "Textures/floor/raw"
PROCESSED_DIR = "Textures/floor/processed"
TARGET_SIZE = (32, 32)
```

```python
os.makedirs(PROCESSED_DIR, exist_ok=True)
```

```python
valid_images = []
rejected_images = []
```

```python
for fname in sorted(os.listdir(RAW_DIR)):
    if not fname.lower().endswith(".png"):
        continue

    input_path = os.path.join(RAW_DIR, fname)
    output_path = os.path.join(PROCESSED_DIR, fname)

    try:
        with Image.open(input_path) as img:
            img = img.convert("RGB")

            if img.size != TARGET_SIZE:
                raise ValueError(f"Invalid size {img.size}")

            img.save(output_path, format="PNG")
            valid_images.append(fname)

    except Exception as e:
        print(f"[REJECTED] {fname}: {e}")
        rejected_images.append(fname)
```

```python
print("\nSUMMARY")
print("--------")
print(f"Valid images   : {len(valid_images)}")
print(f"Rejected images: {len(rejected_images)}")
```
