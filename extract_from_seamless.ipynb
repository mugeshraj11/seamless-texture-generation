```python
import os
import hashlib
from PIL import Image

INPUT_DIR = "Textures/floor/seamless"
OUTPUT_DIR = "Textures/floor/seamless_tiles_32"

TILE_SIZE = 32
seen = set()

os.makedirs(OUTPUT_DIR, exist_ok=True)

count = 0
for fname in os.listdir(INPUT_DIR):
    if not fname.endswith(".png"):
        continue

    img = Image.open(os.path.join(INPUT_DIR, fname)).convert("RGB")

    # take ONE representative tile (top-left)
    tile = img.crop((0, 0, TILE_SIZE, TILE_SIZE))

    # hash to ensure uniqueness
    h = hashlib.md5(tile.tobytes()).hexdigest()
    if h in seen:
        continue

    seen.add(h)
    tile.save(f"{OUTPUT_DIR}/tile_{count:03d}.png")
    count += 1

print("Unique 32x32 tiles saved:", count)
```
